### Task1 : Write a program that counts up the number of vowels [a, e, i, o, u] contained in the string.

```
Task1 = "An English paragraph is a group of sentences or a single sentence that forms a unit, focusing on one main, cohesive idea or controlling idea."
Task1L = Task1.lower()
if "a" or "e" or "i" or "o" or "u" in Task1L:
    print(Task1L.count("a")+Task1L.count("e")+Task1L.count("i")+Task1L.count("o")+Task1L.count("u"))
```

### Task1 : after searching

```
Task1 = "An English paragraph is a group of sentences or a single sentence that forms a unit, focusing on one main, cohesive idea or controlling idea."
Task1L = Task1.lower()
vowel = "aeiou"
count = 0

for char in Task1L:
    if char in vowel:
        count += 1

print(f"There are {count} vowels in the paragraph.")

# .lower(), for, in, count +=1, f-format,
```

### Task3 : Write a program that remove all vowels from  the input word and generate a brief version of it

```
Task3 = input("Please enter A word: ")
print(Task3)
vowels = "aeiouAEIOU"
for char in Task3:
    if char in vowels:
        Task3 = Task3.replace(char,"")
print(Task3)

# input, for, in, if
```

### Task5: Write a program that built a Mario pyramid like below

```
Task5 = input("please enter the height of the Mario pyramid: ")
while not Task5.isdigit():
    Task5 = input("please enter a valid height of the Mario pyramid: ")
row = 1
Task5 = int(Task5)
while row <= Task5:
    spaces = " " * (Task5-row)
    stars = row * "*"
    print (spaces + stars)
    row += 1
```
